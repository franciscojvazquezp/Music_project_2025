```python
## Cleaning the columns name: Convert to lower case, quit spaces and apply snake_case.
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')
## Rename the column 'userid'
df.rename(columns={'userid': 'user_id'}, inplace=True)
## Print the columns to prove the changes made
print(df.columns)
```

```python
## Calculate the number of missing values
print(df.isna().sum())
```

**Missing values:** 
The DataFrame includes missing values for the columns: track, artist and genre. In the next step we are going to fill these values 
with the word 'unknown'.

```python
## Create a list with the columns that include missing values.
columns_to_replace= ['track','artist','genre']
## By using a loop fill the missing values with the word 'unknown'.
for column_to_replace in columns_to_replace:
    df.fillna('unknown', inplace= True)
## Calculate the missing values in order to prove the changes made.
print(df.isna().sum())
```

```python
## Calculate the duplicated values that includes the DataFrame.
print(df.duplicated().sum())
```

**Explicit duplicated values:** <br>
The DataFrame includes 3,826 duplicated values. In the next step we are going to delete those values.

```python
## Delete duplicated values
df = df.drop_duplicates()
## Calculate the duplicated values to prove the changes made.
print(df.duplicated().sum())
```

We are going to examine the list to identify **implicit duplicates** of the hip-hop genre, that are misspelled names or variants that refer to the same musical genre.

The duplicated values that we will find below:

* `hip`  
* `hop`  
* `hip-hop`  

Define a function called 'replace_wrong_values()' with the parameters below:

* `df`: The DataFrame to modify
* `column`: The name of the column used
* `wrong_values`: A list of the wrong values
* `correct_value`: The correct value to replace the wrong

```python
## Function to replace the implicit duplicated values.
def replace_wrong_values(df,column,wrong_values,correct_value):
    for wrong_value in wrong_values:
        df[column]= df[column].replace(wrong_value,correct_value)
    return df
```

```python
## By using the function delete the implicit duplicated values in the column 'genre'.
replace_wrong_values(df,'genre',['hip', 'hop', 'hip-hop'],'hiphop')
```

```python
## Show the list of the unique values in the column 'genre' in order to confirm the changes made.
print(df['genre'].unique())
```

```python
## Show the quantity of unique values in the column 'genre' to confirm the changes made.
print(df['genre'].nunique())
```
